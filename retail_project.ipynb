```python
import pandas as pd

data = {
    "order_id": [1, 2, 3, 3, 4, 5, 6, 7, 8, 9],
    "product": ["Rice", "Sugar", "Oil", "Oil", "Tea", "Rice", "Sugar", None, "Tea", "Oil"],
    "city": ["Lahore", "Karachi", "Lahore", "Lahore", "Multan", "Karachi", "Lahore", "Multan", "Karachi", None],
    "quantity": [5, 2, 3, 3, None, 4, 6, 2, 3, 1],
    "price": [200, 150, 500, 500, 300, 200, 150, 300, 300, 500]
}

df = pd.DataFrame(data)
df.to_csv("sales_raw.csv", index=False)
print(df)
```

```python
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
clean = df.drop_duplicates()
clean = clean.dropna(subset=["product", "city"])
clean["quantity"] = clean["quantity"].fillna(clean["quantity"].median())
clean["total"] = clean["quantity"] * clean["price"]
print(clean)
```

```python
clean.to_csv("sales_clean.csv", index=False)
```

```python
print(clean.groupby("product")["total"].sum().sort_values(ascending=False))
print(clean.groupby("city")["total"].sum())
```
