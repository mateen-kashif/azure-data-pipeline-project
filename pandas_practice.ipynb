```python
import pandas as pd

data = {
    "name": ["Ali", "Sara", "Ahmed", "Fatima"],
    "age": [22, 25, 30, 28],
    "city": ["Lahore", "Karachi", "Multan", "Islamabad"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df[df['age'] > 24])
```

```python
print(df[['name', 'city']])
```

```python
print(df.sort_values('age', ascending=False))
```

```python
print(df[(df['age'] > 24) & (df['city'] == 'Multan')])
```

```python
import pandas as pd

data = {
    "name": ["Ali", "Sara", "Sara", "Ahmed", "Fatima", "Bilal"],
    "age": [22, 25, 25, None, 28, 35],
    "city": ["Lahore", "Karachi", "Karachi", "Multan", None, "Lahore"]
}

df2 = pd.DataFrame(data)
print(df2)
```

```python
print(df2.isnull().sum())
```

```python
df2 = df2.drop_duplicates()
print(df2)
```

```python
print(df2.dropna())
```

```python
print(df2.fillna({"age": 0, "city": "Unknown"}))
```
