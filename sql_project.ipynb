```python
import sqlite3
import pandas as pd

clean = pd.read_csv("sales_clean.csv")
conn = sqlite3.connect("sales.db")
clean.to_sql("sales", conn, if_exists="replace", index=False)

print(pd.read_sql("SELECT * FROM sales", conn))
```

```python
print(pd.read_sql("SELECT SUM(total) AS total_sales FROM sales", conn))
```

```python
print(pd.read_sql("""
SELECT product, SUM(total) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC
""", conn))
```

```python
print(pd.read_sql("""
SELECT city, SUM(total) AS revenue
FROM sales
GROUP BY city
ORDER BY revenue DESC
""", conn))
```

```python
print(pd.read_sql("SELECT * FROM sales WHERE total > 1000", conn))
```

```python
print(pd.read_sql("""
SELECT product, ROUND(AVG(total), 2) AS avg_order
FROM sales
GROUP BY product
""", conn))
```
